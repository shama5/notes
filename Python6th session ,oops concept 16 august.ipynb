## . Encapsulation

Binding the variable(attributes) and methods inside single unit(class) and restricting direct access to the data.

It prevents data from direct modification by making variables private and giving controlled access using metods

It provides data hiding by making variables private and allows access only through public methods, ensuring controlled and secure interaction with the data.

👉 In simple words:
Encapsulation = Putting data in a capsule (class) and giving controlled access using methods.

```python
#Acess Specifiers
# 1. Public -> can be accessed by anyone
# 2. Private -> can be only accessed inside the class and cannot acess outside the class
# 3. protected -> it can be accessed inside the class and inside the subclasses
# __variable name -> 2 underscore before variable for decalaring private variable
```

```python
#Getter and setter
#Getter is used for getting the value, & setter is used for setting the value.We can use any name in place of getter & setter
class bank:
    def __init__(self):
        self.name = "Swastik"
        self.__balance = 10000
    def getter(self):
        print("Bank balance is: ",self.__balance)
    def setter(self):
        self.__balance = 12000
ob = bank()
print(ob.name)
ob.getter()  #private variable can not be accessed directly,but it can be accessed via method
ob.setter()  #setter is only used for changing the value
ob.getter()
```

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance   # private variable (hidden)

    # method to deposit money
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
        else:
            print("Invalid deposit amount")

    # method to withdraw money
    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            print(amount)
            self.__balance -= amount
        else:
            print("Invalid withdrawal")

    # method to check balance
    def get_balance(self):
        return self.__balance
account = BankAccount(1000)   # opening account with 1000

# ✅ Correct way (through methods)
account.deposit(500)
print(account.get_balance())   # Output: 1500

print(account.withdraw(-500))

# ❌ Wrong way (trying to access private variable directly)
print(account.__balance)   # AttributeError: object has no attribute '__balance'


# __balance → private variable (hidden from outside, protected from direct modification).

# deposit(), withdraw(), get_balance() → public methods (controlled way to access/modify balance).

# This ensures data safety → you can’t just set balance = -999 from outside.
```

# Inheritance doesn’t mean Python automatically calls all parent methods.It only means the child class has access to those methods.

Types of Inheritance

1. Single Inheritance : Single parent and single child

2. Multiple Inheritance : Multiple parent and single child

3. Multilevel Inheritance : Grandfather -> father -> child

4. Heirarichal Inheritance : Single parent and multiple child

5. Hybrid Inheritance : Multiple types of inheritance in 1 code.

```python
# I have class Employee
# I want to implement new class named as tech employee
#here calling parent class inside child class using super method
class Employee:
    def __init__(self,name,empid):
        self.name = name
        self.empid = empid
    def displayparent(self):
        print(self.name,self.empid)

class TechEmployee(Employee):
    def __init__(self,name,empid,team):
    #super() -> it a special function that provides a way to access methods of parent class inside child class
        super().__init__(name,empid)
        #We use super without self
        self.team = team
    def displaychild(self):
        super().displayparent()
        print(self.team)

emp = TechEmployee("Swastik",8607,"AI/ML TEAM")
emp.displaychild()
```

# Abstraction
 Hides unnecessary details from the user and shows only what is needed.
It focuses on what an object does instead of how it does it.
Done using abstract classes or interfaces.

```python
from abc import ABC, abstractmethod
# abc ->  Abstraction Base class
# ABC -> CLASS provided by abc
class Vehicle(ABC): #abstract class
    @abstractmethod
    def start(self):
        pass
    @abstractmethod
    def stop(self):
        pass
    def display(self):
        pass
```

```python
#Concrete class - child of abstract class
#it is mandatory to implement all the abstract methods inside concrete class
class Car(Vehicle):
    def start(self):
        print("Started") 
    def stop(self):
        print("ended")
ob = Car()
ob.start()
ob.stop()
```

```python
from abc import ABC, abstractmethod

# Abstract Class
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass   # only declaration, no implementation

    @abstractmethod
    def perimeter(self):
        pass

# Concrete Class
class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

# Concrete Class
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius

    def perimeter(self):
        return 2 * 3.14 * self.radius


# Using abstraction
r = Rectangle(5, 10)
c = Circle(7)

print("Rectangle Area:", r.area())
print("Rectangle perimeter:", r.perimeter())
print("circle Area:", c.area())
print("Circle Perimeter:", c.perimeter())
```

- Shape is an abstract class → only defines what methods must exist (area, perimeter).

- Rectangle and Circle implement those methods → this is the real behavior.

You can’t create an object of Shape directly (it’s abstract).

```python
Store Discount System with Overriding

Story: In a shopping system, every Product has a price. A DiscountedProduct gives 10% off and overrides the method.
```

```python
class Product:
  def __init__(self,item,price):
    self.item = item
    self.price = price
  def get_price(self):
    print("The final price of ",self.item," is ",self.price)

class DiscountedProduct(Product):
  def get_price(self):
    self.price = self.price * 0.9
    print("The final price of ",self.item," is ",self.price)

ob1 = Product("Shoes",1500)
ob1.get_price()
ob2 = DiscountedProduct("Shirt",2000)
ob2.get_price()
ob1 = DiscountedProduct("Shoes",1500)
ob1.get_price()
```

```python
Perfect number
```

```python
def perfect_number(n):
  per = 1
  for i in range(2,n//2 + 1):
    if n%i == 0:
      per += i
  if per==n:
    return True
  else:
    return False
n = int(input())
if perfect_number(n)==True:
  print("It is perfect number")
else:
  print("It is not perfect number")
```

```python
Ice Cream Shop

Create a class IceCream that takes a flavor and has methods to serve and restock.
```

```python
class IceCream:
  def __init__(self,flavour,stock):
    self.flavour = flavour
    self.stock = stock
  def serve(self,quantity):
    if quantity <= self.stock:
      print(quantity," scoops of ",self.flavour," has been served, enjoy🍨")
      self.stock -= quantity
    else:
      print(self.flavour," is out of stock😓")
  def restock(self,quantity):
    self.stock += quantity
    print(quantity," number of scoops of",self.flavour," has been added to stock 😊")
  def checkstock(self):
    print(self.stock," scoops of ",self.flavour," is left🍨")
ob1 = IceCream("Vanilla",10)
ob2 = IceCream("Chocolate",10)
ob3 = IceCream("ButterScotch",10)
ob1.serve(8)
ob2.serve(7)
ob3.serve(6)
ob2.checkstock()
ob1.serve(4)
ob1.restock(10)
ob1.serve(11)
ob1.checkstock()
ob2.checkstock()
ob3.checkstock()
```
