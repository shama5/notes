### Q1. Write a Python program to demonstrate multiple inheritance.
1. Employee class has 3 data members EmployeeID, Gender (String), Salary and PerformanceRating(Out of 5) of type int. It has a get() function to get these details from the user.
2. JoiningDetail class has a data member DateOfJoining of type Date and a function getDoJ to get the Date of joining of employees.
3. Information Class uses the marks from Employee class and the DateOfJoining date from the JoiningDetail class to calculate the top 3 Employees based on their Ratings and then Display, using readData, all the details on these employees in Ascending order of their Date Of Joining.



```python
from datetime import datetime

# Employee Class
class Employee:
    def __init__(self):
        self.EmployeeID = ""
        self.Gender = ""
        self.Salary = 0
        self.PerformanceRating = 0

    def get(self):
        self.EmployeeID = input("Enter Employee ID: ")
        self.Gender = input("Enter Gender (M/F): ")
        self.Salary = int(input("Enter Salary: "))
        self.PerformanceRating = int(input("Enter Performance Rating (out of 5): "))


class JoiningDetail:
    def __init__(self):
        self.DateOfJoining = None  # initially empty

    def getDoJ(self):
        date_str = input("Enter Date of Joining (YYYY-MM-DD): ")
        self.DateOfJoining = datetime.strptime(date_str, "%Y-%m-%d").date()


# Information class inherits from Employee and JoiningDetail
class Information(Employee, JoiningDetail):
    def __init__(self):
        Employee.__init__(self)
        JoiningDetail.__init__(self)

    def readData(self):
        print("Employee ID:", self.EmployeeID)
        print("Employee ID:",self.EmployeeID)
        print("Gender:", self.Gender)
        print("Salary:",self.Salary)
        print("Performance Rating:",self.PerformanceRating)
        print("Date of Joining:",self.DateOfJoining.strftime('%Y-%m-%d'))


n = int(input("Enter number of employees: "))
employees = []

for i in range(n):
    print(f"\nEnter details for Employee #{i + 1}")
    emp = Information()
    emp.get()
    emp.getDoJ()
    employees.append(emp)

# Sort by performance rating descending
top_employees = sorted(employees, key=lambda x: x.PerformanceRating, reverse=True)[:3]

# Sort top 3 by date of joining ascending
top_employees.sort(key=lambda x: x.DateOfJoining)

# Display result
print("\nTop 3 Employees based on Performance Rating (sorted by Date of Joining):")
for emp in top_employees:
    emp.readData()
```

### Q.2 Write a Python program to demonstrate Polymorphism.
1. Class Vehicle with a parameterized function Fare, that takes input value as fare and returns it to calling Objects.
2. Create five separate variables Bus, Car, Train, Truck and Ship that call the Fare function.
3. Use a third variable TotalFare to store the sum of fare for each Vehicle Type.
4. Print the TotalFare.



```python
# Class definition
class Vehicle:
    def Fare(self, fare):
        return fare

# Create instances of Vehicle for each type
Bus = Vehicle()
Car = Vehicle()
Train = Vehicle()
Truck = Vehicle()
Ship = Vehicle()

# Each vehicle calls the same method with its own fare
bus_fare = Bus.Fare(100)
car_fare = Car.Fare(150)
train_fare = Train.Fare(300)
truck_fare = Truck.Fare(250)
ship_fare = Ship.Fare(500)

# Polymorphism: Same method, different objects, different values
TotalFare = bus_fare + car_fare + train_fare + truck_fare + ship_fare

# Output the result
print("Total Fare collected from all vehicles is:", TotalFare)
```

```python
#this is the most appropriate way
# Base class
class Vehicle:
    def Fare(self):
        return 0  # default (can also raise NotImplementedError)

# Subclasses with overridden Fare method
class Bus(Vehicle):
    def Fare(self):
        return 100

class Car(Vehicle):
    def Fare(self):
        return 150

class Train(Vehicle):
    def Fare(self):
        return 300

class Truck(Vehicle):
    def Fare(self):
        return 250

class Ship(Vehicle):
    def Fare(self):
        return 500

# Create instances of each subclass
vehicles = [Bus(), Car(), Train(), Truck(), Ship()]

# Polymorphism: call the same method on different objects
TotalFare = 0
for v in vehicles:
    TotalFare += v.Fare()

# Print the result
print("Total Fare collected from all vehicles is:", TotalFare)
```
