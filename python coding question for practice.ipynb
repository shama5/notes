##### Remove Vowels from String in Python

```python
s = "Hello World"

result = ""

for i in s:
    if i not in "aeiouAEIOU":
        result = result + i

print(result)
```

```python
s = "Hello World"

result = "".join([i for i in s if i not in "aeiouAEIOU"])

print(result)
```

```python
#Find Frequency of Characters
s = "apple"

freq = {}

for i in s:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1

print(freq)
```

```python
#count Uppercase and Lowercase
s = "PyThOn"

upper = 0
lower = 0

for i in s:
    if i.isupper():
        upper += 1
    else:
        lower += 1

print("Upper:", upper)
print("Lower:", lower)
```

```python
#Separate Even and Odd Numbers
nums = [1,2,3,4,5,6]

even = []
odd = []

for i in nums:
    if i % 2 == 0:
        even.append(i)
    else:
        odd.append(i)

print("Even:", even)
print("Odd:", odd)
```
